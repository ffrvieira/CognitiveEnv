```python
#importação bibliotecas
import boto3
```

```python
textract = boto3.client('textract', region_name='us-east-1')
```

```python
def extrair_texto_documento(caminho_imagem):
    with open(caminho_imagem, 'rb') as img_file:
        resposta = textract.detect_document_text(Document={'Bytes': img_file.read()})
    textos = [item["DetectedText"] for item in resposta["Blocks"] if item["BlockType"] == "LINE"]
    return textos
```

```python
# Exemplo de uso
textos = extrair_texto_documento('data/exemplo_documento.jpg')
print(textos)
```
